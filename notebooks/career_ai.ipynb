```python
# ============================================
# STEP 2: Imports & Project Configuration
# ============================================

import os
import sys
from pathlib import Path

import pandas as pd
import numpy as np

# RAG / Document processing
import chromadb
from sentence_transformers import SentenceTransformer
from pypdf import PdfReader

# Environment variables
from dotenv import load_dotenv

# Display basic information
print("✅ All required libraries imported successfully!")
print("Python:", sys.version)
print("ChromaDB:", chromadb.__version__)

# ============================================
# Project Paths
# ============================================

PROJECT_ROOT = Path.cwd()

DATA_DIR = PROJECT_ROOT / "data"
UPLOAD_DIR = DATA_DIR / "uploads"
CHROMA_DIR = PROJECT_ROOT / "chroma_db"

# Create directories if they don't exist
UPLOAD_DIR.mkdir(parents=True, exist_ok=True)
CHROMA_DIR.mkdir(parents=True, exist_ok=True)

print("\n📁 Project Root:")
print(PROJECT_ROOT)

print("\n📂 Upload Directory:")
print(UPLOAD_DIR)

print("\n🗄️ ChromaDB Directory:")
print(CHROMA_DIR)

# ============================================
# Load environment variables
# ============================================

load_dotenv()

print("\n🔐 Environment configuration loaded.")
```

```python
# ============================================
# STEP 3: Load Embedding Model
# ============================================

EMBEDDING_MODEL_NAME = "all-MiniLM-L6-v2"

print("🔄 Loading embedding model...")
print("Model:", EMBEDDING_MODEL_NAME)

embedding_model = SentenceTransformer(EMBEDDING_MODEL_NAME)

print("✅ Embedding model loaded successfully!")
print("📐 Embedding dimension:", embedding_model.get_sentence_embedding_dimension())
```

```python
# ============================================
# STEP 4: Initialize ChromaDB
# ============================================

print("🔄 Initializing ChromaDB...")

# Persistent ChromaDB client
chroma_client = chromadb.PersistentClient(
    path=str(CHROMA_DIR)
)

# Create or load collection
collection = chroma_client.get_or_create_collection(
    name="career_documents",
    metadata={
        "description": "CareerAI document knowledge base"
    }
)

print("✅ ChromaDB initialized successfully!")
print("📚 Collection:", collection.name)
print("📄 Documents currently stored:", collection.count())
```

```python
# ============================================
# STEP 5: PDF Text Extraction
# ============================================

def extract_pdf_text(pdf_path):
    """
    Extract text from a PDF file.
    """
    reader = PdfReader(str(pdf_path))

    pages_text = []

    for page_number, page in enumerate(reader.pages, start=1):
        text = page.extract_text()

        if text and text.strip():
            pages_text.append({
                "page": page_number,
                "text": text.strip()
            })

    return pages_text


print("✅ PDF extraction function created!")
```

```python
# ============================================
# STEP 6: Text Chunking
# ============================================

def chunk_text(text, chunk_size=800, overlap=150):
    """
    Split text into overlapping chunks.
    """

    text = text.strip()

    if not text:
        return []

    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        start += chunk_size - overlap

    return chunks


print("✅ Text chunking function created!")
```

```python
# ============================================
# STEP 7: Process Uploaded PDFs
# ============================================

def process_pdf(pdf_path):
    """
    Extract text from PDF and split it into chunks.
    """

    pages = extract_pdf_text(pdf_path)

    all_chunks = []

    for page_data in pages:
        chunks = chunk_text(page_data["text"])

        for chunk in chunks:
            all_chunks.append({
                "text": chunk,
                "page": page_data["page"],
                "source": Path(pdf_path).name
            })

    return all_chunks


pdf_files = list(UPLOAD_DIR.glob("*.pdf"))

print(f"📄 PDFs found: {len(pdf_files)}")

if pdf_files:
    for pdf in pdf_files:
        print("   •", pdf.name)
else:
    print("ℹ️ No PDFs uploaded yet.")
    print("   You can upload career/resume/job-related PDFs later.")
```

```python
# ============================================
# STEP 8: Create Embeddings + Store in ChromaDB
# ============================================

def add_documents_to_vector_db(chunks):
    """
    Convert text chunks into embeddings and store them in ChromaDB.
    """

    if not chunks:
        print("⚠️ No chunks to store.")
        return

    documents = [item["text"] for item in chunks]

    # Create unique IDs for every chunk
    ids = [
        f"chunk_{collection.count() + i}"
        for i in range(len(documents))
    ]

    # Create embeddings for all chunks
    print(f"🔄 Creating embeddings for {len(documents)} chunks...")

    embeddings = embedding_model.encode(
        documents,
        show_progress_bar=True
    ).tolist()

    # Store documents + embeddings + useful metadata
    metadatas = [
        {
            "source": item["source"],
            "page": str(item["page"])
        }
        for item in chunks
    ]

    collection.add(
        ids=ids,
        documents=documents,
        embeddings=embeddings,
        metadatas=metadatas
    )

    print("✅ Documents successfully stored in ChromaDB!")
    print("📚 Total chunks in database:", collection.count())
```

```python
# ============================================
# STEP 9: Process PDFs and Store in ChromaDB
# ============================================

pdf_files = list(UPLOAD_DIR.glob("*.pdf"))

print(f"📄 Found {len(pdf_files)} PDF file(s)\n")

if not pdf_files:
    print("⚠️ No PDF found.")
    print("Please put a PDF inside: data/uploads/")
else:
    for pdf_path in pdf_files:
        print(f"🔄 Processing: {pdf_path.name}")

        chunks = process_pdf(pdf_path)

        print(f"   📑 Created {len(chunks)} chunks")

        add_documents_to_vector_db(chunks)

        print(f"   ✅ Stored successfully\n")

print("================================")
print("📚 Total chunks in ChromaDB:", collection.count())
print("================================")
```

```python
# ============================================
# STEP 10: Semantic Search / Retrieval
# ============================================

def retrieve_documents(query, top_k=3):
    """
    Retrieve the most relevant document chunks
    from ChromaDB for a user query.
    """

    # Convert query into embedding
    query_embedding = embedding_model.encode(
        query
    ).tolist()

    # Search ChromaDB
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=top_k
    )

    retrieved_documents = results["documents"][0]
    retrieved_metadata = results["metadatas"][0]
    distances = results["distances"][0]

    return retrieved_documents, retrieved_metadata, distances


# Test query
query = "What skills and technologies are mentioned in the resume?"

documents, metadata, distances = retrieve_documents(query)

print("🔎 QUERY:")
print(query)

print("\n" + "=" * 60)

for i, (doc, meta, distance) in enumerate(
    zip(documents, metadata, distances), start=1
):
    print(f"\n📌 RESULT {i}")
    print(f"Source: {meta['source']}")
    print(f"Page: {meta['page']}")
    print(f"Distance: {distance:.4f}")
    print("-" * 50)
    print(doc[:1000])
```

```python
# Step 11: Gemini-powered RAG answer generation

from google import genai
from dotenv import load_dotenv
import os

# Load .env
load_dotenv()

# Check API key
api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
    raise ValueError("GEMINI_API_KEY not found in .env file.")

# Create Gemini client
client = genai.Client(api_key=api_key)

print("✅ Gemini client initialized successfully!")


def generate_answer(question, top_k=4):

    # -----------------------------------
    # 1. Create embedding for user query
    # -----------------------------------
    query_embedding = embedding_model.encode(
        [question]
    ).tolist()[0]

    # -----------------------------------
    # 2. Retrieve relevant chunks
    # -----------------------------------
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=top_k
    )

    retrieved_documents = results["documents"][0]
    retrieved_metadatas = results["metadatas"][0]

    # -----------------------------------
    # 3. Build context
    # -----------------------------------
    context_parts = []

    for doc, metadata in zip(
        retrieved_documents,
        retrieved_metadatas
    ):
        source = metadata.get("source", "Unknown")
        page = metadata.get("page", "Unknown")

        context_parts.append(
            f"[Source: {source}, Page: {page}]\n{doc}"
        )

    context = "\n\n".join(context_parts)

    # -----------------------------------
    # 4. Create RAG prompt
    # -----------------------------------
    prompt = f"""
You are CareerAI, an intelligent career assistant.

Answer the user's question using the provided context.

IMPORTANT RULES:
- Use the context as the primary source.
- Do not invent information about the user's resume.
- If the information is not present in the context, clearly say that it is not available.
- Give practical and concise career-oriented answers.
- Make the answer easy to read.

CONTEXT:
{context}

USER QUESTION:
{question}

ANSWER:
"""

    # -----------------------------------
    # 5. Ask Gemini
    # -----------------------------------
    response = client.models.generate_content(
        model="gemini-3.8-flash",
        contents=prompt
    )

    return response.text
```

```python
# Step 12: Test CareerAI RAG

question = "What technical skills does Yuvraj have?"

answer = generate_answer(question)

print("=" * 70)
print("🤖 CareerAI")
print("=" * 70)
print("\nQuestion:")
print(question)

print("\nAnswer:")
print(answer)

print("\n" + "=" * 70)
```
